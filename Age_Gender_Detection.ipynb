````python
def get_completion(prompt: str, model="gpt-4-turbo"):
        # say the role of model
        messages = [
            {"role": "system", "content": "suppose you're gender classifier model"},
            {"role": "user", "content": prompt},
        ]
        t0 = time()
        response = openai.ChatCompletion.create(
            model=model,
            messages=messages,
            temperature=0,  # this is the degree of randomness of the model's output
            max_tokens=650,
        )
        t1 = time()
        usage = response.usage
        usage = {**usage, "time": f"{t1-t0:.2f} seconds"}
        print(usage)

        return response.choices[0].message["content"]
````

````python
# Prompt Template
prompt_template = """
You are an expert gender classifier. Your task is to classify the gender of these names into one of three categories: "female", "male" or "unknown".
```
{comment}
```
Your response should be one of the following labels: female, male, unknown.
"""
````

````python
import os
from PIL import Image
import cv2

# Function to convert an image to JPG format and resize it
def convert_and_resize_image(image_path, output_path, size=(400, 400)):
    with Image.open(image_path) as img:
        img = img.convert('RGB')  # Convert to RGB mode
        img = img.resize(size)    # Resize image
        img.save(output_path, 'JPEG')

# Function to check if a face is in the photo
def is_face_in_photo(image_path):
    # Load the pre-trained Haar Cascade classifier for face detection
    face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

    # Read the image
    img = cv2.imread(image_path)
    if img is None:
        print(f"Failed to load image: {image_path}")
        return False

    # Convert the image to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Detect faces in the image
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

    # Check if faces are detected
    if len(faces) > 0:
        print(f"Faces detected in {image_path}: {faces}")
        return True
    else:
        print(f"No face detected in {image_path}")
        return False

# Directory where the photos are stored
photos_directory = '/content/all_images'  # Replace with the actual directory containing the photos
converted_directory = '/content/converted_images'  # Directory to save converted images

# Create the directory if it does not exist
if not os.path.exists(converted_directory):
    os.makedirs(converted_directory)

# List all files in the photos directory
files = os.listdir(photos_directory)

# Convert and resize all files to JPG format
for file_name in files:
    image_path = os.path.join(photos_directory, file_name)
    # Generate new file name for JPG
    new_file_name = os.path.splitext(file_name)[0] + '.jpg'
    output_path = os.path.join(converted_directory, new_file_name)
    try:
        convert_and_resize_image(image_path, output_path)
    except Exception as e:
        print(f"Failed to convert and resize {image_path} to JPG: {e}")

# Initialize the results list with NaNs
results = [np.nan] * 2903  

# Process each converted file
converted_files = os.listdir(converted_directory)
for file_name in converted_files:
    # Check if the file name is a number and within the specified range
    if file_name.split('.')[0].isdigit() and 0 <= int(file_name.split('.')[0]) <= 2902:
        index = int(file_name.split('.')[0])
        image_path = os.path.join(converted_directory, file_name)
        # Check for face in the photo
        if is_face_in_photo(image_path):
            results[index] = 'face'
        else:
            results[index] = 'no-face'
    else:
        print(f"Skipping file: {file_name}")

# Create a list of photos that do not exist in the results
not_exist_list = [index for index, result in enumerate(results) if pd.isna(result)]

# Output the results
print("Results List:", results)
print("Photos that do not exist in the list:", not_exist_list)

# Count 'face' values in the results list
face_count = results.count('face')
print(f"Number of photos with faces detected: {face_count}")

df = pd.DataFrame(results, columns=['face_detect'])

print(df)
````

# age detection for all photos

````python
from deepface import DeepFace
import pandas as pd
import numpy as np

photos_dir = '/content/all_images/all_images' 
output_dir = '/content/resized_images'

os.makedirs(output_dir, exist_ok=True)

# Process all images in the directory
for filename in os.listdir(photos_dir):
    if filename.lower().endswith(('.jpeg', '.jpg', '.png')):  # Process JPEG and PNG files
        img_path = os.path.join(photos_dir, filename)
        img = cv2.imread(img_path)
        if img is not None:
            resized_img = cv2.resize(img, (400, 400))
            new_filename = os.path.splitext(filename)[0] + '.jpg'  # Convert to .jpg
            new_img_path = os.path.join(output_dir, new_filename)
            cv2.imwrite(new_img_path, resized_img)

age_list = []

for index, row in df.iterrows():
    if row['face_detect'] == 'face':
        photo_filename = f"{index}.jpg"  
        photo_path = os.path.join(output_dir, photo_filename)
        try:
            analysis = DeepFace.analyze(img_path=photo_path, actions=['age'])
            formatted_output = analysis[0] if len(analysis) > 0 else {'age': np.nan, 'face_confidence': np.nan}
            age_list.append({'age': formatted_output['age'], 'face_confidence': formatted_output['face_confidence']})
        except Exception as e:
            print(f"Error analyzing sample image {photo_filename}: {e}")
            age_list.append({'age': np.nan, 'face_confidence': np.nan})  # Handle error by setting age and face_confidence to NaN
    else:
        age_list.append({'age': np.nan, 'face_confidence': np.nan})  # Handle case where 'face_detect' is not 'face'

df_result = pd.DataFrame(age_list, columns=['age', 'face_confidence'])

print(df_result)
````
